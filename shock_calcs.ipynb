```python
#figure out if i'm f'd with 4kg/mm front springs

car_weight = 995 #kg
car_f2b_weight_ratio = 2/5 #front to back weight ratio (40:60)
driver_weight = 105 #kg
driver_f2b_weight_ratio = 1/2 
total_weight = car_weight + driver_weight
front_weight = car_weight*car_f2b_weight_ratio + driver_weight*driver_f2b_weight_ratio
rear_weight =  car_weight*(1-car_f2b_weight_ratio) + driver_weight*(1-driver_f2b_weight_ratio)

```

```python
coilover_length = 0 #mm
total_stroke = 200 #mm
bump_stop_length = 25.4 #mm

front_spring_constant = 4 #kg/mm
rear_spring_constant = 8 #kg/mm
spring_length = 200 #mm
preload_length = 5 #mm
front_preload = preload_length*front_spring_constant
rear_preload = preload_length*rear_spring_constant

print(f"{front_preload = :.2f}kg; {rear_preload = :.2f}kg")

def calc_sag(corner_weight, preload_length, spring_constant):
    sag = (corner_weight - preload_length*spring_constant)/spring_constant #mm
    return sag

def mm2in(mm):
    inch = mm/25.4
    return inch

front_sag = calc_sag(front_weight/2, preload_length, front_spring_constant)
remaining_front_stroke = total_stroke-front_sag-bump_stop_length-preload_length
rear_sag = calc_sag(rear_weight/2, preload_length, rear_spring_constant)
remaining_rear_stroke = total_stroke-rear_sag-bump_stop_length-preload_length

print(f"{front_weight = :.2f}kg, front_corner_weight = {front_weight/2 :.2f}kg;")
print(f"{rear_weight = :.2f}kg, rear_corner_weight = {rear_weight/2 :.2f}kg")
print(f"spring_length = {total_stroke:.2f}mm, {mm2in(total_stroke):.2f}in")
print(f"{front_sag = :.2f}mm, {mm2in(front_sag):.2f}in; {remaining_front_stroke = :.2f}mm")
print(f"{rear_sag = :.2f}mm, {mm2in(rear_sag):.2f}in; {remaining_rear_stroke = :.2f}mm")
print(f"{mm2in(remaining_front_stroke) = :.2f}in; {mm2in(remaining_rear_stroke) = :.2f}in")









```
